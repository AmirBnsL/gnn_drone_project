# Sinkhorn Head — Fine-tuning

Loads the trained `LocalNegotiatorGNN` checkpoint and fine-tunes it with a
differentiable `SinkhornHead` that replaces the gossip consensus inference.

### What changes vs the gossip training

| | Gossip (train_strict_decentralized) | Sinkhorn (this notebook) |
|---|---|---|
| Inference | gossip consensus (non-differentiable) | Sinkhorn iterations (differentiable) |
| Loss | CE + conflict penalty + regret | soft CE + regret (no conflict term — Sinkhorn guarantees it) |
| Bijection | ~0.50 ceiling | guaranteed by Hungarian rounding of P |
| Gradient path | encoder only | encoder + Sinkhorn head end-to-end |

Use the `torch_env` kernel.

```python
import os, random
import numpy as np
import torch

from local_negotiator import (
    COMM_RADIUS,
    NUM_FORMATIONS,
    SLOT_VISIBILITY_RADIUS,
    MAX_CONSENSUS_ROUNDS,
    LocalNegotiatorGNN,
    evaluate_strict_decentralized,
    load_negotiator_dataset,
    prepare_dataset,
)
from sinkhorn_head import (
    SinkhornHead,
    SINKHORN_ITERS,
    evaluate_sinkhorn,
    train_sinkhorn_model,
)

# ── Config ────────────────────────────────────────────────────────────────────
SEED             = 42
DATASET          = './dataset/negotiator_dataset_v1.pt'
BASE_CKPT        = 'strict_local_negotiator_best.pt'   # your trained gossip checkpoint
OUT              = 'sinkhorn_best.pt'

EPOCHS           = 60
LR               = 3e-4
FREEZE_EPOCHS    = 15    # encoder frozen for first N epochs
MAX_TRAIN        = 7000
MAX_VAL          = 500
MAX_TEST         = 500
MIN_BIJ_FOR_BEST = 0.80  # higher bar than gossip training (0.75)

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f'Device : {DEVICE}')
print(f'Torch  : {torch.__version__}')
```

```python
# ── Load dataset ──────────────────────────────────────────────────────────────
if not os.path.isfile(DATASET):
    raise FileNotFoundError(f'Dataset not found at {DATASET}')
if not os.path.isfile(BASE_CKPT):
    raise FileNotFoundError(
        f'Base checkpoint not found at {BASE_CKPT}. '
        'Run train_strict_decentralized.ipynb first.'
    )

# Build model first so we can use its formation_embedding for dataset prep
model = LocalNegotiatorGNN().to(DEVICE)

# Load the gossip-trained weights into the base model
ckpt = torch.load(BASE_CKPT, map_location='cpu', weights_only=False)
model.load_state_dict(ckpt['model_state_dict'])
model.to(DEVICE)
print('Loaded base checkpoint:')
print(f"  gossip bijection (test) : {ckpt['metrics']['test']['bijection_rate']:.3f}")
print(f"  gossip cost ratio (test): {ckpt['metrics']['test']['cost_ratio_vs_hungarian']:.4f}")
print(f"  gossip match (test)     : {ckpt['metrics']['test']['slot_match_rate']:.3f}")

# Prepare dataset using the loaded (trained) formation embeddings
raw_list = load_negotiator_dataset(DATASET)
train_data, val_data, test_data = prepare_dataset(
    raw_list,
    model.formation_embedding.weight.detach(),
    seed=SEED,
    force_gt_visibility=False,
)
train_data = train_data[:MAX_TRAIN]
val_data   = val_data[:MAX_VAL]
test_data  = test_data[:MAX_TEST]
print(f'\nTrain={len(train_data)}  Val={len(val_data)}  Test={len(test_data)}')
```

```python
# ── Sinkhorn head ─────────────────────────────────────────────────────────────
sinkhorn_head = SinkhornHead(n_iters=SINKHORN_ITERS).to(DEVICE)
print(f'SinkhornHead params    : {sum(p.numel() for p in sinkhorn_head.parameters())}')
print(f'Base model params      : {sum(p.numel() for p in model.parameters()):,}')
print(f'Sinkhorn log_temp init : {sinkhorn_head.sinkhorn_log_temp.item():.4f}')
print(f'Encoder frozen for     : first {FREEZE_EPOCHS} epochs')
print(f'Then joint training at : lr × 0.3 = {LR * 0.3:.2e}')
```

```python
# ── Baseline: gossip inference on test set (for comparison) ───────────────────
print('=== Baseline: gossip inference (no Sinkhorn) ===')
gossip_test = evaluate_strict_decentralized(model, test_data, DEVICE)
for k, v in gossip_test.items():
    print(f'  {k}: {v:.4f}')
```

```python
# ── Train ─────────────────────────────────────────────────────────────────────
print('=== Sinkhorn fine-tuning ===')
history = train_sinkhorn_model(
    model,
    sinkhorn_head,
    train_data,
    val_data,
    DEVICE,
    epochs=EPOCHS,
    lr=LR,
    freeze_epochs=FREEZE_EPOCHS,
    min_bijection_for_best=MIN_BIJ_FOR_BEST,
    force_gt_train=True,
)
```

```python
# ── Final evaluation — Sinkhorn vs gossip on the same test set ────────────────
print('=== Final evaluation ===')

sink_train = evaluate_sinkhorn(model, sinkhorn_head, train_data[:200], DEVICE)
sink_val   = evaluate_sinkhorn(model, sinkhorn_head, val_data,         DEVICE)
sink_test  = evaluate_sinkhorn(model, sinkhorn_head, test_data,        DEVICE)

print('\nSinkhorn inference:')
print(f'  Train: {sink_train}')
print(f'  Val  : {sink_val}')
print(f'  Test : {sink_test}')

print('\nGossip inference (same model, same test set):')
gossip_final = evaluate_strict_decentralized(model, test_data, DEVICE)
print(f'  Test : {gossip_final}')

print('\n=== Head-to-head comparison (test set) ===')
keys = ['bijection_rate', 'cost_ratio_vs_hungarian', 'slot_match_rate',
        'conflict_rate', 'unassigned_rate']
print(f'{"metric":<35} {"gossip (base)":>16} {"sinkhorn":>12}')
print('-' * 65)
for k in keys:
    g = gossip_test.get(k, 0.0)
    s = sink_test.get(k, 0.0)
    arrow = '↑' if (k != 'cost_ratio_vs_hungarian' and s > g) or \
                   (k == 'cost_ratio_vs_hungarian' and s < g) else ' '
    print(f'{k:<35} {g:>16.4f} {s:>12.4f} {arrow}')
```

```python
# ── Save checkpoint ───────────────────────────────────────────────────────────
torch.save(
    {
        'model_state_dict':   model.state_dict(),
        'sinkhorn_state_dict': sinkhorn_head.state_dict(),
        'config': {
            'num_formations':         NUM_FORMATIONS,
            'comm_radius':            COMM_RADIUS,
            'slot_visibility_radius': SLOT_VISIBILITY_RADIUS,
            'max_consensus_rounds':   MAX_CONSENSUS_ROUNDS,
            'sinkhorn_iters':         SINKHORN_ITERS,
            'freeze_epochs':          FREEZE_EPOCHS,
            'base_checkpoint':        BASE_CKPT,
        },
        'history': history,
        'metrics': {
            'sinkhorn': {'train': sink_train, 'val': sink_val, 'test': sink_test},
            'gossip_baseline': gossip_test,
        },
    },
    OUT,
)
print(f'Saved → {OUT}')
```

```python
# ── Training curves ───────────────────────────────────────────────────────────
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
fig.suptitle('Sinkhorn fine-tuning', fontsize=13, fontweight='bold')

ep = range(1, len(history['train_loss']) + 1)

axes[0].plot(ep, history['train_loss'], label='train', color='steelblue')
axes[0].plot(ep, history['val_loss'],   label='val',   color='coral', linestyle='--')
axes[0].axvline(FREEZE_EPOCHS, color='gray', linestyle=':', alpha=0.7, label='unfreeze')
axes[0].set_title('Loss'); axes[0].legend(); axes[0].grid(alpha=0.3)

axes[1].plot(ep, history['val_bijection_rate'],          label='bijection', color='green')
axes[1].plot(ep, history['val_slot_match_rate'],         label='match',     color='purple', linestyle='--')
axes[1].axvline(FREEZE_EPOCHS, color='gray', linestyle=':', alpha=0.7)
axes[1].axhline(0.80, color='green', linestyle=':', alpha=0.5, label='target bij')
axes[1].set_title('Bijection & Match Rate'); axes[1].legend(); axes[1].grid(alpha=0.3)

axes[2].plot(ep, history['val_cost_ratio_vs_hungarian'], label='cost ratio', color='orange')
axes[2].axvline(FREEZE_EPOCHS, color='gray', linestyle=':', alpha=0.7)
axes[2].axhline(1.0, color='gray', linestyle='-', alpha=0.3, label='Hungarian optimal')
axes[2].set_title('Cost Ratio vs Hungarian'); axes[2].legend(); axes[2].grid(alpha=0.3)

plt.tight_layout()
plt.savefig('sinkhorn_training_curves.png', dpi=120)
plt.show()
print('Saved sinkhorn_training_curves.png')
```
